```python
import pandas as pd
from pathlib import Path

csv_content = """student_id,name,department,age,marks
1001,Aarav,Computer Science,20,86
1002,Diya,Electronics,21,91
1003,Kabir,Mechanical,20,74
1004,Ananya,Computer Science,22,95
1005,Ishaan,Civil,21,68
1006,Meera,Electronics,20,88
1007,Rohan,Mechanical,22,79
1008,Sara,Computer Science,21,93
1009,Vivaan,Civil,20,72
1010,Aditi,Electronics,22,84
1011,Arjun,Mechanical,21,77
1012,Nisha,Computer Science,20,90
"""

Path("dataset.csv").write_text(csv_content, encoding="utf-8")

print("Dataset created successfully.")
```

```python
df = pd.read_csv("dataset.csv")

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("FIRST FIVE RECORDS")
display(df.head())
```

```python
print("LAST FIVE RECORDS")
display(df.tail())
```

```python
print("COLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES")
print(df.dtypes)

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDATASET SHAPE")
print(df.shape)
```
